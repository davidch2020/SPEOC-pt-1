```python
import pandas as pd

# set paths
input_csv = "final_data_CD.csv"

# read CSV - base CSV is already totalled by person
df = pd.read_csv(input_csv, usecols=["Group Name", "unpaid_interest_adj", "final_total_adj"], low_memory=False)

# create additional dataframes to remove zeros for respective graphs
df_unpaid_interest_filtered = df[df["unpaid_interest_adj"] != 0]
df_final_total_filtered = df[df["final_total_adj"] != 0]
```

```python
# zoomed graph for final_total (6p & 6p_def)
import matplotlib.pyplot as plt

# sort by final_total_adj
sorted_data_6p = df_final_total_filtered.sort_values("final_total_adj", ascending=False).reset_index()

# set axis
x=range(len(sorted_data_6p))
y=sorted_data_6p["final_total_adj"]

# plot
plt.figure(figsize=(12, 6))
plt.step(x, y, where="mid", color="black", linewidth=1)
plt.fill_between(x, y, step='mid', alpha=0.3, color="lightsteelblue")

#label x axis
percentiles = [75, 50, 25]
percentile_positions = [int(((100-p)) / 100 * (len(sorted_data_6p) - 1)) for p in percentiles]
percentile_labels = [f"P{p}" for p in percentiles]
plt.xticks(percentile_positions, percentile_labels)
plt.xlabel("Percentile of Debt Held")
plt.xlim(0, len(sorted_data_6p))

# label y axis
plt.ylabel("Total 6% & Deferred 6% Stock")
plt.ylim(0,)


# names to search
treasurer_names = [
    "Gerard Bancker",
    "Jedediah Huntington",
    "John Haywood",
    "John Meals",
    "William Gardner",
    "William Richardson"
]

# create vertical bars & labels
highlight_indices = sorted_data_6p[sorted_data_6p["Group Name"].isin(treasurer_names)].index.tolist()
for idx in highlight_indices:
    name = sorted_data_6p.loc[idx, "Group Name"]
    value = y[idx]
    plt.vlines(idx, 0, value, color='red', linewidth=1.5)
    plt.text(idx, value + 1000, name, rotation=45, ha="left", va="bottom", fontsize=8)

# final formatting & show
plt.title("Total 6% & Deferred 6% Stock per Person")
plt.tight_layout()
plt.show()
```

```python
# calculate and print percentile ranks for selected treasurers
print("Treasurer Percentile Ranks:\n")

treasurer_names = [
    "Gerard Bancker",
    "Jedediah Huntington",
    "John Haywood",
    "John Meals",
    "William Gardner",
    "William Richardson"
]

for name in treasurer_names:
    match = sorted_data_6p[sorted_data_6p["Group Name"] == name]
    if not match.empty:
        idx = match.index[0]
        percentile = 100 * idx / (len(sorted_data_6p) - 1)
        sp_value = match["final_total_adj"].values[0]
        print(f"{name:<25} — Index: {idx:<4}  Percentile: {100-percentile:.2f}%  6p Value: {sp_value:,.2f}")
    else:
        print(f"{name:<25} — Not found in data.")
```
